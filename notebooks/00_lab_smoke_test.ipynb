# AI_RESEARCH_LAB smoke test

Этот ноутбук проверяет, что пакет импортируется, результаты demo_study доступны, а артефакты и Logbook существуют.

```python
from pathlib import Path
import json
import sys
import pandas as pd
import matplotlib.pyplot as plt

def find_project_root(start: Path | None = None) -> Path:
    start = start or Path.cwd()
    for candidate in [start, *start.parents]:
        if (candidate / 'config' / 'lab.yaml').exists() and (candidate / 'research').exists():
            return candidate
    raise FileNotFoundError('Не удалось найти корень AI_RESEARCH_LAB')

project_root = find_project_root()
project_root
```

```python
from ai_research_lab import __version__
from ai_research_lab.paths import study_paths

study = study_paths('demo_study', base_root=project_root)
print('Python:', sys.version.split()[0])
print('Package version:', __version__)
print('Study root:', study.root)
print('Logbook exists:', (study.logbook / 'LOGBOOK.md').exists())
print('Artifacts registry exists:', (study.logs / 'artifact_registry.json').exists())
```

```python
metrics_path = study.results / 'metrics.json'
metrics = json.loads(metrics_path.read_text(encoding='utf-8'))
summary = pd.DataFrame(metrics['summary_by_sample_size'])
summary
```

```python
raw_csv = study.results / 'raw_results.csv'
raw_df = pd.read_csv(raw_csv)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(summary['sample_size'], summary['mean_abs_error'], marker='o')
ax.set_xscale('log')
ax.set_xlabel('Sample size')
ax.set_ylabel('Mean absolute error')
ax.grid(True, which='both', linestyle='--', alpha=0.4)
fig.tight_layout()
fig
```
